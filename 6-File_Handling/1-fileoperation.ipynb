File Operation- Read and Write Files

File handling is a crucial part of any programming language. Python provided built-in functions and methods to read from and write to files, both text and binary. This lesson will cover the basics of file handling, including reading aand writing text files and binary files.

```python
## Read a whole file

with open('example.txt', mode='r') as file:
    content = file.read()
    print(content)
```

```python
## read the file line by line

with open('example.txt', mode='r') as file:
    for line in file:
        print(line.strip())  ## strip() removes the newline character

```

```python
## write a file (with overwriting)

with open('example.txt', mode='w') as file:
    file.write('Hey this is an example of overwriting\n')
    file.write('this is a new line')
```

```python
## write a file( without overwriting) using append mode

with open('example.txt', mode='a') as file:
    file.write('append operation taking place!\n')
```

```python
## writing a list of lines to a file

lines = ['first word\n','second word\n', 'third word\n']

with open('example.txt', mode='a') as file:
    file.writelines(lines)
```

```python
## binary files
# in binary files we usually update bytes 

# write to a binary file  

# note rb - read byte and wb - write byte 

data = b'\x00\x01\x02\x03\x04'
with open('example.bin', mode='wb') as file:
    file.write(data)
```

```python
## to read the binary file

with open('example.bin', mode='rb') as file:
    print(file.read())
```

```python
## Read the content from a source text file and write to a destination text file

with open('source.txt', mode='r') as file:
    content = file.read()

with open('destination.txt', mode='w') as file:
    file.write(content)
    
```

```python
## Read a text file and count the number of lines, words, and characters.

with open('source.txt', mode='r') as file:
    
    lines = file.readlines()

    num_of_lines = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)
    print(num_of_lines)
    print(word_count)
    print(char_count)
```

The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exits, its content is truncated (i.e., the file is overwritten).

```python
## writing and then reading a file

with open('source.txt', mode='w+') as fp:
    fp.write("Hello world!\n")
    fp.write("This is my second line\n")


    ## move the file cursor to the begining
    fp.seek(0)

    ## read the content of file
    content = fp.read()
    print(content)
```
