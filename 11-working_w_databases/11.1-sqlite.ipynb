SQL and SQLite

SQL(Structutred Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained , serverless and zero-configuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite , including creating databases, tables, and performing various SQL operations.

```python
import sqlite3
```

```python
## Connect to the sqlite database

connection = sqlite3.connect('example.db')

connection

```

```python

## cursor is basically used to iterate through each rows and columns.
# A cursor is an essential tool when working with databases, 
# as it is used to send commands over to the database and retrieve data when requested.

cursor = connection.cursor()
```

```python
## creating a table

cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    ) 
''' )

## commit the changes
connection.commit()

```

```python
cursor.execute('''
Select * From employees
               
''')
```

```python
## Insert the data

cursor.execute('''
Insert Into employees(name, age, department)
               values ('Pranav', 24, 'Data Science')

''')

cursor.execute('''
Insert Into employees(name, age, department)
               values ('Amit', 25, 'Finance')

''')

cursor.execute('''
Insert Into employees(name, age, department)
               values ('Sourav', 30, 'Arts')

''')

## for commit
connection.commit()
```

```python
# query the data from the table

cursor.execute('Select * From employees')
rows = cursor.fetchall()

## print all the queried data

for row in rows:
    print(row)
```

```python
## update the data in the table

cursor.execute('''
Update employees
Set name="Megha"
where id = '1'
''')

connection.commit()
```

```python
cursor.execute('Select * From employees')
rows = cursor.fetchall()

## print all the queried data

for row in rows:
    print(row)
```

```python
## delete the data from the table

cursor.execute('''
Delete from employees
where name = "Pranav"
''')

connection.commit()
```

```python
cursor.execute('Select * From employees')
rows = cursor.fetchall()

## print all the queried data

for row in rows:
    print(row)
    
```

```python
## one example to understand all the things

connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

## creating table

cursor.execute('''
Create Table If Not Exists sales (
               id Integer Primary Key,
               date Text Not Null,
               product Text Not Null,
               sales Integer,
               region Text
               )
''')


## Insert the data in the table

# cursor.execute('''
# Insert into sales (date,product,sales, region)               
#                 values ('2025-02-01', 'Paper Lite', '1000', 'US')
# '''
# )

# cursor.execute('''
# Insert into sales (date,product,sales, region)               
#                 values ('2025-02-05', 'Paper Standard', '3000', 'India')
# '''
# )

# cursor.execute('''
# Insert into sales (date,product,sales, region)               
#                 values ('2025-02-10', 'Paper Premium', '5000', 'China')
# '''
# )

# you also can do this 
sales_data = [
    ('2025-02-01', 'PaperLite', 1000, 'US'),
    ('2025-02-05', 'PaperStandard', 3000, 'India'),
    ('2025-02-10', 'PaperPremium', 5000, 'China')
]


## Insert the data in the table
# this is for bulk insertion 

cursor.executemany('''
Insert into sales(date, product, sales, region)
                   values (?,?,?,?)
''', sales_data)


connection.commit()
```

```python
cursor.execute('Select * From sales')
rows = cursor.fetchall()

## print all the queried data

for row in rows:
    print(row)
```

```python
## close the connection

connection.close()

# this is the last step 
```
