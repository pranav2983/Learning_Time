Pandas - DataFrame and Series

Pandas is a powerful data manipulation library in python,,widely used for data analysis and data cleaning. It provides primarly two data structures : Series and Dataframes. A series is one-dimensional array like object. while a DataFrame is a two-dimensional, size-mutuable , and  potentially heterogenous tabular data structure with labeled axes(rows and columns).

```python
## series
## a pandas series is a one-dimensional array-like object that can hold any data type. it is similar to a column in a table.

import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series", series)
print(type(series))
```

```python
## create a series from dictonary
data = {'a':1,'b':2,'c':3}
pd.Series(data)

```

```python
data = [10,20,30]
index = ['a','b','c']

pd.Series(data, index = index)


## so without using any key-values pair , how do i need to use this index over here

```

```python
## DataFrame
## create a Dataframe from a dictionary of a list

data = {
    'Name':['Pranav','Raj', 'Sheena'],
    'Age': ['21', '20', '21'],
    'City': ['Bangalore', 'New York', 'Delhi']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# if i want to convert this in an array

import numpy as np

np.array(df)
```

```python
## create a dataframe from a list of dictonary

data = [
    {'Name':'Pranav','Age': 32 , 'City' : 'Bangalore'},
    {'Name':'Avnish','Age': 20 , 'City' : 'Delhi'},
    {'Name':'Rutik','Age': 24 , 'City' : 'Pune'},
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
