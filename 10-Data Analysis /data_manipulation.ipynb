Data manipulation and Analysis with Pandas

data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform and extract insights from data. In this lession , we will cover various data manipulation and analysis techniques using Pandas.

```python
## if there is any missing values and youy have to fill then use
df_filled = df.fillna(0)
print(df_filled)
```

```python
## filled missing values with the mean of the column
df['Sales_fillNA'] = df['facecream'].fillna(df['facecream'].mean())
df
```

```python
df.dtypes
```

```python
# renaming the columns

df = df.rename(columns={'month_number':'Month'})
df
```

```python
## changing the datatypes

df['New_facecream'] = df['facecream'].astype(float)
df
```

```python
## apply function in any column

df['New_facewash'] = df['facewash'].apply(lambda x:x/100)
df.head()
```

```python
## Data Aggregating and Grouping

df.head()
```

```python
df['Product']=['A','B','C','D','E','F','G','H','I','J','K','L']
df
```

```python
grouped_mean = df.groupby('Product')['facewash'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['Product', 'Month'])['total_units'].sum()
print(grouped_sum)
```

```python
grouped_meann = df.groupby(['Product', 'Month'])['total_units'].mean()
print(grouped_meann)
```

```python
## aggregate multiple functions

grouped_agg = df.groupby('Product')['total_profit'].agg(['mean', 'sum', 'count'])
grouped_agg
```

```python
## merging and joining dataframes


## create sample DataFrames

df1 = pd.DataFrame({'key' : ['A','B','C'], 'Value1' : [1,2,3]})
df2 = pd.DataFrame({'key' : ['A', 'B', 'D'], 'Value2' : [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
## Merge dataframe on the 'Key Columns'

pd.merge(df1, df2, on='key', how='inner')
```

```python
pd.merge(df1, df2, on='key', how='outer')
```

```python
pd.merge(df1, df2, on='key', how='left')
```

```python
pd.merge(df1, df2, on='key', how='right') ## right side (df2) will have high priority
```
