```python
# this is the correct way to defining the attr. and methods 
# in the first hand 

## __init__ -> If initially i want to define any attribute, 
# i can make sure that with the help of this particular constructor 
# at the first hand i'll be able to define all the attributes that will 
# be avaiable inside this particular class.


## Instance variable
class Dog:
    # __init__  method is called constructor

    def __init__(self, name, age):

        # name and age are my parameter and internally i will go ahead 
        # and create some instance variable. Now this instance variable are like
        # some kind of properties that are avaible inside this dog class 
        
        self.name=name #instance variable or properties
        self.age=age


## create the object

dog1 = Dog("shimba", 2)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2 = Dog("chick", 2)
print(dog2)
```

```python
# define a class with instance methods
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof")

dog1=Dog('buddy', 3)
dog1.bark()

dog2=Dog('lucy', 2)
dog2.bark()
```

Example to Understand it more

```python
## Modeling a bank account

## Define a class for bank account

class Account:
    
    def __init__(self,owner, balance=0):
        #defining instance variables
        self.owner = owner
        self.balance=balance

    # definig instance methods
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited and your new balance is {self.balance} ")

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. Now your new balance is {self.balance}")

    def get_balance(self):
        return self.balance
    

# create an object
account1 = Account('Pranav', 2300)
print(account1.balance)

```

```python
# call instance methods
account1.deposit(5700)
```

```python
## withdrew some money

account1.withdraw(3000)
```

```python
print(account1.get_balance())
```
