```python
# 1

def read_file(filename):
    with open(filename, mode='r') as file:
        for line in file:
            print(line.strip())


read_file('sample.txt')
```

```python
# 2 

def write_file(lines, filename):
    with open(filename, mode='w') as file:
        for line in lines:
            file.write(line + '\n')


write_file(['hello','world'], 'output.txt')
```

```python
def copy_file(source_filename, destination_filename):
    with open(source_filename, mode='r') as src:
        with open(destination_filename, mode='w') as des:
            des.write(src.read())

copy_file('source.txt', 'destination.txt')
```

```python
def append_str(str, filename):
    with open(filename, mode='a') as app:
        app.write(str)

append_str('hello', 'log.txt')
```

```python
def count_words(filename):
    with open(filename, mode='r') as file:
        lines = file.readlines()
        print(lines) 
        count_words = sum(len(line.split()) for line in lines)
        print(count_words)

count_words('log.txt')
```
